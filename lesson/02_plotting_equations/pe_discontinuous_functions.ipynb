# Practical exercise: plotting discontinuous functions

For the following, write up the formula in $\LaTeX$ and then depict it on a Cartesian plane. Identify and annotate up any discontinuities and any limits associated with those discontinuities.

 1. The Heaviside Step Function *(function $f:\mathbb{R} \to \{0,\frac{1}{2},1\}$, which takes a value zero for inputs less than 0, 1 for inputs greater than 0, and one half at zero)*
 2. The ratio between $x^3$ and $x^2 - 2 x + 1$.
 3. The ratio between $x^2 - 1$ and $x - 1$
 4. The tangent function.
